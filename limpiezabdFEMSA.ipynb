```python
import pandas as pd  #Base de datos y datos iniciales
df= pd.read_excel("Base de Operaciones Credito TEC.xlsx", sheet_name= "Operaciones")
df.head()
```

```python
df.shape # Cantidad de filas (1009) y cantidad de columnas (18)
```

```python
columns_text = df.select_dtypes(include="object").columns
for col in columns_text:
    df[col]=df[col].str.strip()
    # Quitamos espacios innecesarios al inicio o final del nombre de las columnas
```

```python
print("Filas duplicadas encontradas:", df.duplicated().sum())
df = df.drop_duplicates()
```

```python
print(df.isnull().sum())
#Comprobamos que no haya columnas con valores nulos 
```

```python
df.dropna()
```

```python
for col in ["Solicitud", "Fecha_Deposito", "Fecha_de_primer_pago"]:
    df[col] = pd.to_datetime(df[col])
    #Corroborar formato fechas
```

```python
df["Status_cod"] = df["Status"].map({"Activo": 0, "Liquidado": 1})
df["Frecuencia_cod"] = df["Frecuencia"].map({"Semanal": 0, "Mensual": 1})
#Mapeo de variables por categoria
```

```python
salida = "Base_Operaciones_CreditoFEMSA_LIMPIA.xlsx"
df.to_excel(salida, index=False)
```
